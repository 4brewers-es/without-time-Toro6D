Este notebook valida la coherencia interna del marco TORO6, derivando más de 10 constantes fundamentales a partir de solo dos axiomas geométricos ($Γ = e/5$ y $\delta = 2/\pi$).

```python
import numpy as np
import math
from scipy.optimize import fsolve

# Define koide_eq for fsolve to work
def koide_eq(x):
    # m_mu_pred debe estar en el global scope tras su calculo
    return 2*(1 + m_mu_pred + x) - 3*(1 + np.sqrt(m_mu_pred) + np.sqrt(x))**2

print("--- TORO6 v9 · AUDITORÍA COMPLETA ---\n")

# AXIOMAS
Gamma = math.e / 5.0
Delta_crit = 2.0 / math.pi
Gap = Delta_crit - Gamma
N_moduli = 42

# CONSTANTE DE ESTRUCTURA FINA (Leakage)
inv_alpha_obs = 137.035999
alpha_obs = 1 / inv_alpha_obs
alpha_geo = Gap / (4 * math.pi)
inv_alpha_geo = 1 / alpha_geo
leakage_factor = (alpha_obs - alpha_geo) / alpha_obs

# CASCADA DE MASAS (Electron)
M_Planck_GeV = 1.22e19 
S_E = 2 * math.pi * N_moduli * (1 + 2 * abs(leakage_factor))
v_geo = M_Planck_GeV * math.exp(-S_E / 7.0)
m_e_pred = (v_geo * 1000) * (alpha_obs**2) / 20
m_e_obs = 0.510998

# MUÓN Y TAUÓN
m_mu_pred = 1.5 * inv_alpha_geo + 4
m_tau_pred = fsolve(koide_eq, 3500)[0]
ratio_tau_mu = m_tau_pred / m_mu_pred
ratio_obs = 1776.86 / 105.658367

# HUBBLE
H0_Planck = 67.4
H0_local = H0_Planck * (1 + Gap / (2 * Gamma))

# MATERIA OSCURA
epsilon_edgb = Gamma * Gap
R_dm = (1 / (2 * Gap)) * (1 - epsilon_edgb)
R_dm_obs = 5.11

print(f"[1] Electrón (MeV): {m_e_pred:.6f} (Obs: {m_e_obs:.6f}) -> {100*m_e_pred/m_e_obs:.3f}%")
print(f"[2] Muón (me): {m_mu_pred:.6f} (Obs: 206.768) -> {100*m_mu_pred/206.768:.4f}%")
print(f"[3] Tauón Ratio: {ratio_tau_mu:.6f} (Obs: {ratio_obs:.6f}) -> {100*ratio_tau_mu/ratio_obs:.4f}%")
print(f"[4] Materia Oscura: {R_dm:.4f} (Obs: {R_dm_obs}) -> {100*R_dm/R_dm_obs:.2f}%")
print(f"[5] H0 Local (Pred): {H0_local:.2f} km/s/Mpc (Obs: 73.04) -> {100*H0_local/73.04:.2f}%")
```
